```python
import matplotlib.pyplot as plt
import numpy as np

image_data = model.get("OTRS:IN20:571:Image:ArrayData")
q_nominal = model.get("QUAD:IN20:525:BCTRL")

# scan quad +/- 10%
q_scan = np.linspace(q_nominal * 0.0, q_nominal * 1.1, 5)

for q in q_scan:
    model.set({"QUAD:IN20:525:BCTRL": q})
    image_data = model.get("OTRS:IN20:571:Image:ArrayData")

    plt.figure()
    plt.imshow(np.array(image_data[400:600, 600:800]))
```

```python
fig, ax = plt.subplots()
ax.plot(model.get("stat:mean_z"), model.get("stat:sigma_x"), label="sigma_x")
ax.plot(model.get("stat:mean_z"), model.get("stat:sigma_y"), label="sigma_y")
ax.plot(model.get("stat:mean_z"), model.get("stat:sigma_z"), label="sigma_z")
ax.set_xlabel("Mean Z")
ax.set_ylabel("Sigma")
ax.legend()
```
