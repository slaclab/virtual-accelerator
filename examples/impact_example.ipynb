```python
import matplotlib.pyplot as plt
import numpy as np

image_data = model.get_value("OTRS:IN20:571:Image:ArrayData")
q_nominal = model.get_value("QUAD:IN20:525:BCTRL")

# scan quad +/- 10%
q_scan = np.linspace(q_nominal * 0.0, q_nominal * 1.1, 5)

for q in q_scan:
    model.set({"QUAD:IN20:525:BCTRL": q})
    image_data = model.get_value("OTRS:IN20:571:Image:ArrayData")

    plt.figure()
    plt.imshow(np.array(image_data[400:600, 600:800]))
```
